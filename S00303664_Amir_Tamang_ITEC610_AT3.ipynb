# **Task 1**

```python
import pandas as pd  # Importing the pandas library for data manipulation
import matplotlib.pyplot as plt  # Importing matplotlib for data visualization

# Load the dataset from the specified CSV file
data = pd.read_csv('/content/ITEC610_assessment3_data.csv')

# Calculate summary statistics for the specified columns: weight2, wtyrago (weight a year ago), and htm3 (height)
summary = data[['weight2', 'wtyrago', 'htm3']].describe()
print("Summary Statistics:\n", summary)  # Print the summary statistics for verification

# Define category labels for the plot
categories = ['Current_weight', 'Weight_a_year_ago', 'Height']

# Extract statistical values for each category
min_vals = summary.loc['min'].values  # Minimum values
max_vals = summary.loc['max'].values  # Maximum values
mean_vals = summary.loc['mean'].values  # Mean values
median_vals = summary.loc['50%'].values  # Median values (50th percentile)
std_vals = summary.loc['std'].values  # Standard deviation values
percentile_25 = summary.loc['25%'].values  # 25th percentile values
percentile_75 = summary.loc['75%'].values  # 75th percentile values

# Plotting summary statistics
plt.figure(figsize=(8, 6))  # Set the figure size
plt.scatter(categories, max_vals, marker='v', color='black', label='max', s=100)  # Maximum values
plt.scatter(categories, min_vals, marker='^', color='black', label='min', s=100)  # Minimum values
plt.scatter(categories, mean_vals, marker='x', color='blue', label='mean', s=100)  # Mean values
plt.scatter(categories, median_vals, marker='*', color='red', label='median', s=100)  # Median values
plt.scatter(categories, mean_vals + std_vals, marker='v', color='green', label='mean+std', s=100)  # Mean + standard deviation
plt.scatter(categories, mean_vals - std_vals, marker='^', color='green', label='mean-std', s=100)  # Mean - standard deviation
plt.scatter(categories, percentile_75, marker='>', color='orange', label='75th percentile', s=100)  # 75th percentile
plt.scatter(categories, percentile_25, marker='<', color='orange', label='25th percentile', s=100)  # 25th percentile

plt.legend(loc='upper right')  # Add legend to the upper right corner
plt.title("Summary Statistics of Current Weight, Weight a Year Ago, and Height")  # Add plot title
plt.ylabel("Values (kg/cm)")  # Label the y-axis
plt.tight_layout()  # Adjust layout for better fit
plt.show()  # Display the plot
```

# **Task 2**

```python
# Calculate the change in weight by subtracting the weight from a year ago from the current weight
data['weight_change'] = data['weight2'] - data['wtyrago']

# Create a dictionary to store the correlations between weight change and other variables
correlations = {
    'weight2': data['weight_change'].corr(data['weight2']),  # Correlation between weight change and current weight
    'wtyrago': data['weight_change'].corr(data['wtyrago']),  # Correlation between weight change and weight a year ago
    'age': data['weight_change'].corr(data['age'])           # Correlation between weight change and age
}

# Create a scatter plot to visualize the relationship between weight change and current weight
plt.scatter(data['weight_change'], data['weight2'])  # X-axis: Weight change, Y-axis: Current weight
plt.title('Weight Change vs Current Weight')  # Add a title to the plot
plt.xlabel('Weight Change')  # Label the X-axis
plt.ylabel('Current Weight')  # Label the Y-axis
plt.show()  # Display the plot

# Create another scatter plot to visualize weight change against weight from a year ago
plt.scatter(data['weight_change'], data['wtyrago'])  # X-axis: Weight change, Y-axis: Weight a year ago
plt.title('Weight Change vs Weight a Year Ago')  # Add a title to the second plot
plt.xlabel('Weight Change')  # Label the X-axis
plt.ylabel('Weight a Year Ago')  # Label the Y-axis
plt.show()  # Display the second plot

# Create a third scatter plot to visualize weight change against age
plt.scatter(data['weight_change'], data['age'])  # X-axis: Weight change, Y-axis: Age
plt.title('Weight Change vs Age')  # Add a title to the third plot
plt.xlabel('Weight Change')  # Label the X-axis
plt.ylabel('Age')  # Label the Y-axis
plt.show()  # Display the third plot
```

# **Task 3**

```python
from sklearn.linear_model import LinearRegression  # Import the LinearRegression class from sklearn

# Data: Defining house sizes (in square feet) and their corresponding prices (in dollars)
house_sizes = [[1500], [2000], [2500], [3000], [3500]]  # House sizes as a 2D list (required for sklearn)
house_prices = [250000, 300000, 330000, 360000, 390000]  # List of house prices

# Model: Creating an instance of the LinearRegression model
model = LinearRegression()

# Fitting the model with the provided data (training the model)
model.fit(house_sizes, house_prices)

# Prediction: Using the trained model to predict the price of a house with 2500 sqft
predicted_price = model.predict([[2500]])  # Make a prediction for a 2500 sqft house

# Output: Displaying the predicted price in a formatted string
print(f'Predicted price for 2500 sqft house: ${predicted_price[0]:.2f}')  # Print the predicted price with two decimal places

```
